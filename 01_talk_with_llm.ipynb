## Completion Model

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_google_genai import GoogleGenerativeAI

_ = load_dotenv(find_dotenv())
genai_api_key = os.environ["GENAI_API_KEY"]

llm = GoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=genai_api_key
)

response = llm.invoke("Explain Python in one sentence.")
print(response)
```

## Streaming Output

```python
for chunk in llm.stream(
    "Tell me one fun fact about the kennedy Family. I need detailed response"
):
    print(chunk, end="", flush=True)
```

# Configuration of LLM (Temperature etc)

```python
creativeLlm = GoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=genai_api_key,
    temperature=1.9
)

# google has started to ignore temperature in 3.5 or upper models. Default is 1.

response = creativeLlm.invoke("Explain Python in one sentence.")
print(response)
```

# Chat Completion Model or Chat Model

```python
from langchain_google_genai import ChatGoogleGenerativeAI

_ = load_dotenv(find_dotenv())
genai_api_key = os.environ["GENAI_API_KEY"]

chatllm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=genai_api_key
)

messages=[
    ("system","You are an historian expert in the kennedy family."),
    ("human","Tell me one curious thing about JFK")
]

response= chatllm.invoke(messages)
```

```python
print(response.content)
```

```python
response.response_metadata
```

```python
print(response.model_json_schema())
```

# Old langchain

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages=[
    SystemMessage(content="You are an historian expert in the kennedy family."),
    HumanMessage(content="Tell me one curious thing about JFK.")
]

response= chatllm.invoke(messages)

print(response.content)
```

# Open Source LLM (GROQ)

```python
import os
from dotenv import load_dotenv, find_dotenv
_= load_dotenv(find_dotenv())
```

```python
from langchain_groq import ChatGroq

llamaChatModel= ChatGroq(
    model="openai/gpt-oss-120b"
)
```

```python
messages=[
    ("system","You are an historian expert in the Kennedy family."),
    ("human","How many members of the family died tragically?")
]
```

```python
llamaResponse = llamaChatModel.invoke(messages)
print(llamaResponse.content)
```
